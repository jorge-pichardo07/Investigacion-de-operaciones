````python
from sympy import *
````

Se nombran las variables como el ejercicio visto en clase con symbols:

````python
p0, p1, p2, p3,p4= symbols('p0 p1 p2 p3 p4')
````

Se hace en forma matricial el sistema de ecuaciones, en donde la condición de igualdad a 1 la sustituimos por cualquier ecuación, en este caso fue la ecuación 4:

````python
M = Matrix([[-0.7, 0.4, 0.2, 0.33, 0.5,0],
            [0.5, -0.65,0.2, 0,  0 ,0],
            [0.2, 0, -0.8, 0.66, 0, 0],
            [  0, 0.05, 0.2, -1, 0, 0],
            [  1,    1,   1,  1, 1, 1]])
````

````python
M
````

````python
s = solve_linear_system(M, p0, p1, p2, p3, p4)
````

Al usar el comando
```python

solve_linear_system(Matriz, Variables)

```

Se obtiene que:

````python
s
````

Planteamos la ecuación de el número esperado de personas en la fila:

$$
Lq = \sum_{n=1}^4 (n-1)P_n
$$

````python
lq = 0*p0+1*p2 + 2*p3 + 3*p4
````

Entonces $Lq$ queda como:

````python
lq.subs(s)
````
