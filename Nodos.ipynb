# **Arbol de expansión mínima** 

```python
#Agregaré aristas con el nombre del nodo y el peso de la arista
aristas = [
    (1, 2, 1),
    (1, 5, 9),
    (1, 3, 5),
    (2, 5, 3),
    (2, 4, 4),
    (2, 3, 6),
    (1, 4, 7),
    (4, 3, 5),
    (4, 6, 3),
    (3,6,10),
    (4,5,8)
        
]
```

```python
#Grafo
F = nx.Graph()
for u, v, w in aristas:
    F.add_edge(u, v, weight=w)
```

```python
# Dibujar el grafo con pesos en las aristas
pos = nx.spring_layout(F, seed=5)  # Posiciones de los nodos
nx.draw(F,pos, with_labels=True, node_color='blue', edge_color='gray')
```

```python
#Árbol de expansión mínima
mst = nx.minimum_spanning_tree(F)
```

```python
print("Árbol de expansión mínima:")
print(mst.edges(data=True))
print("Peso total del árbol de expansión mínima:", mst.size(weight='weight'))
nx.draw(mst,pos, with_labels=True, node_color='blue', edge_color='red', font_color='white', font_weight='bold')
```
