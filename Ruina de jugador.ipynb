```python
import numpy as np
import matplotlib.pyplot as plt
import random as r
```

```python
prob = [0.6,0.4]
S = [0,1,2,3,4,5]
```

```python
def jugador(P,S):
    lista =[]
    x_0 = 2
    lista.append(x_0) #Porque empezamos en 2
    x_n = x_0

    while  (x_n < S[-1]) and (x_n > S[0]):
        R = r.choices(P) 
        if R == [0.6]:
            x_n += 1
        else:
            x_n -= 1

        lista.append(x_n)
    else:
        pass
    
    return lista
```

```python
M = jugador(prob,S)
plt.plot(M)

    
```
